```python
# Calculating Missing values percentage 
def missing_summary(df, dataset_name):

  missing_count = df.isnull().sum()
  missing_percentage = (missing_count / len(df)) * 100

  summary = pd.DataFrame({
      "Missing_Count": missing_count,
        "Missing_Percentage": missing_percentage.round(2)
    })

  print("\n", dataset_name)
  display(summary)

missing_summary(appointments_df, "Appointments")
missing_summary(clinics_df, "Clinics")
missing_summary(follow_up_df, "Follow-up Notes")
```

```python
# Visualize missing values

missing_counts = appointments_df.isnull().sum()
missing_counts = missing_counts[missing_counts > 0]
missing_counts.plot(
    kind="bar",
    figsize=(9, 5),
    title="Missing Values in Appointments Dataset"
)

plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)
plt.show()
```

```python
# Find appointment IDs in follow-up notes

note_appointment_ids = set(
    follow_up_df["appointment_id"].dropna().unique()
)

# Find appointment IDs in appointments
main_appointment_ids = set(
    appointments_df["appointment_id"].dropna().unique()
)

# Identify unmatched follow-up notes
unmatched_notes = note_appointment_ids - main_appointment_ids
print("Unmatched follow-up appointment IDs:", unmatched_notes)
```
